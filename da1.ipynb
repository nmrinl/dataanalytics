### `if` / `else` — letting the code make decisions

An `if` statement runs some code **only when a condition is true**. `else` covers everything else.

The structure is always: `if <condition>:` then an **indented** block of code underneath. The indentation (usually 4 spaces) is how Python knows which lines belong to the `if`.

```python
score = 90

if score >= 50:
    print("Pass")
else:
    print("Fail")
    print("Study Hard")
```

```python
score = 25
if score != 25:
  print("Found")
else:
  print("Not found")
```

Try changing `score` to a number below 50 in the cell above and re-run it (Shift + Enter). Notice how the output changes.

You can chain more conditions with `elif` (short for "else if") to check several cases in order, top to bottom — Python stops at the first one that's true:

```python
score = 82

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 50:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} => Grade {grade}")
```

The structure looks like: `for <name> in <list>:` followed by an **indented** block — `<name>` becomes each item, one at a time, as the loop runs.

```python
for i in range(100):
  print(i)
```

## 3. Plotting the data

Numbers in a table are hard to "see" patterns in. A plot fixes that.

We'll use `matplotlib`, and make a **scatter plot**: one dot per student, positioned by how many hours they studied (x-axis) and how many practice tests they took (y-axis). We'll color each dot by whether they passed.

```python
import matplotlib.pyplot as plt

# Split the data into two groups so we can color them differently
passed = df[df["passed_exam"] == "Yes"]
failed = df[df["passed_exam"] == "No"]

plt.figure(figsize=(7, 5))
plt.scatter(passed["hours_studied"], passed["practice_tests_taken"],
            color="green", label="Passed", s=80)
plt.scatter(failed["hours_studied"], failed["practice_tests_taken"],
            color="red", label="Failed", s=80)

plt.xlabel("Hours studied")
plt.ylabel("Practice tests taken")
plt.title("Did the student pass?")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. Modifying the data

Now let's use our rule to make a **prediction** for every student. We'll use a `for` loop, just like we practiced earlier, to build up a new list one student at a time.

```python
# Create an empty list to store results
predicted = []

# Loop through each value in the column
for h in df["hours_studied"]:
    if h >= threshold:
        predicted.append("Yes")
    else:
        predicted.append("No")

# Assign the list as a new column
df["predicted_pass"] = predicted

# View result
df.head(10)
```

```python
# Add another column: did our prediction match the real result?
df["correct_prediction"] = df["predicted_pass"] == df["passed_exam"]

accuracy = df["correct_prediction"].mean() * 100
print(f"Our simple rule got {accuracy:.0f}% of students right!")

df.head(10)
```

## 6. A trickier rule: only the top 25% pass

So far our "pass" rule was a **fixed number** — e.g. "5.5 hours or more". Sometimes, though, the rule isn't a fixed number at all.

A common example: an instructor grades **on a curve**, and decides that only the **top 25% of students, by exam score**, will pass — no matter what the actual marks are. Let's work through why that's a different kind of problem.

We'll use a fresh, simple dataset: just a list of 20 students' exam marks (out of 100).

```python
import numpy as np

exam_scores = [55, 62, 47, 78, 91, 68, 73, 85, 59, 64,
               72, 90, 45, 81, 77, 69, 58, 88, 95, 63]

print("Number of students:", len(exam_scores))
print("Marks (highest to lowest):", sorted(exam_scores, reverse=True))
```

### 🕵️ Find the cutoff yourself

Let's plot the marks **sorted from highest to lowest**, against each student's rank (1 = top scorer). With 20 students, the top 25% is the top 5 students.

Look at the plot below and try to guess: **roughly what mark does a student need to be in that top 5?**

```python
sorted_scores = sorted(exam_scores, reverse=True)
ranks = list(range(1, len(sorted_scores) + 1))

plt.figure(figsize=(7, 5))
plt.plot(ranks, sorted_scores, marker="o")
plt.xlabel("Rank (1 = highest mark)")
plt.ylabel("Exam mark")
plt.title("Exam marks, sorted from highest to lowest")
plt.grid(True, alpha=0.3)
plt.show()
```



With 20 students, the top 25% is the top **5** students (25% of 20 = 5). Looking at rank 5 on the plot, the mark there is 85.

image.png

```python

cutoff_score = 85
# Apply the rule with a for loop, same pattern as before
exam_pass = []

for s in exam_scores:
    if s >= cutoff_score:
        exam_pass.append("Yes")
    else:
        exam_pass.append("No")

print(exam_pass)
print(f"Number passing: {exam_pass.count('Yes')} out of {len(exam_scores)}")
```

### The problem: a fixed cutoff doesn't stay correct

We found `cutoff_score` by looking at **this particular group of 20 students**. That number only describes the top 25% *for this group*. Watch what happens when a new batch of students joins the class — say, a stronger set of students transfers in.

```python
# This new batch happens to be a stronger group of students overall
new_students_scores = [82, 89, 95, 91, 87, 84, 93, 99, 90, 86]
all_scores = exam_scores + new_students_scores

print("Total students now:", len(all_scores))
```

```python
# Re-use the OLD fixed cutoff_score we found earlier
old_rule_pass = [s for s in all_scores if s >= cutoff_score]

print(f"Using the OLD fixed cutoff ({cutoff_score}), {len(old_rule_pass)} out of {len(all_scores)} students pass.")
print(f"That's {len(old_rule_pass) / len(all_scores) * 100:.0f}% of the class — not 25% anymore!")
```

```python
sorted_scores = sorted(all_scores, reverse=True)
ranks = list(range(1, len(sorted_scores) + 1))

plt.figure(figsize=(7, 5))
plt.plot(ranks, sorted_scores, marker="o")
plt.xlabel("Rank (1 = highest mark)")
plt.ylabel("Exam mark")
plt.title("Exam marks, sorted from highest to lowest")
plt.grid(True, alpha=0.3)
plt.show()
```

The class size changed, the mix of marks changed, but our rule was still stuck at the same old number — so it no longer picks out the top 25%. We'd have to **manually re-check the plot and re-pick a number every single time** the data changes. That's fragile, and easy to forget to do.

### The data-driven fix: percentiles

A **percentile** answers a question like: *"what mark is the value below which 75% of scores fall?"* That's exactly the same thing as saying *"the top 25% of scores are at or above this mark."*

The key difference from what we did before: instead of us reading a number off a plot and typing it into the code, we ask the computer to **calculate it fresh from whatever data we currently have** — with one line, using `numpy`'s `percentile` function.

```python
import numpy as np
percentile_75 = np.percentile(all_scores, 75)
print(f"75th percentile of the (now bigger) dataset: {percentile_75:.1f}")

new_pass = []

for s in all_scores:
    if s >= percentile_75:
        new_pass.append(s)

print(f"Using the percentile rule: {len(new_pass)} out of {len(all_scores)} pass "
      f"({len(new_pass) / len(all_scores) * 100:.0f}%)")
```

**The takeaway:** whenever the "right answer" depends on the shape of the data itself — like "top 25%", "bottom 10%", or "above average" — hard-coding a fixed number is fragile, because it silently goes stale the moment the data changes. A percentile-based rule recalculates itself from whatever data you feed it, so it keeps meaning the same thing ("top 25%") no matter how many students there are or what they scored. This is a small taste of a much bigger idea in data analysis: **prefer rules that are computed from the data over rules you hard-code by hand.**
