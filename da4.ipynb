# Hypothesis Testing

## What is it?

- Hypothesis testing is a statistical technique used to make informed decisions about a parameter based on a sample of data.

- Decisions can be around mean of a population, proportion of a feature in a population, comparison between 2 different populations.

## Beliefs : Null and alternate hypothesis

<img src="https://images.cartoonstock.com/lowres_800/law-order-innocent-guilty-judge-court-courtroom-rman2371_low.jpg" width="350" height="270">

Checks a random batch of 100 bottles off the assembly line and finds mean volume 501.2 ml. Is it possible that these observations are due to chance, or do they indicate a problem?

* Null hypothesis: current belief about the parameter being evaluated, **the observations are due to random chance**.

* Alternate hypothesis : there is a real effect, **observations are due to this effect + random chance**.

* Similar to the justice example, the null hypothesis is believed to be correct unless proved otherwise.

### Our example


    Null Hypothesis (H0): The mean volume of liquid in the bottles is exactly 500 mL (μ=500).
    Alternative Hypothesis (HA): The mean volume of liquid in the bottles is not 500 mL (μ≠500).

## How does it work?

- Define the problem statement and hypothesis to be tested.

- Find the correct test to perform for the problem at hand.

- Decide the significance level $\alpha$ for the experiment.

- Calculate the test statistic and $p$ value according to the test.

- If $p < \alpha$ then we reject the null hypothesis.

### Wrong choice of significance level $\alpha$

**The chance that we reject $H_0$ when it is actually true (aka Type I error)**

**(1-$\alpha$) is our confidence that any alarm bells we hear are genuine**

- $\alpha$ ia a cutoff point below which we agree that an effect is statistically significant

- Significance level has to be chosen according to the problem and how critical is the application

- 5% and 1% (for more critical applications) are common values

## What is this p value?

**The chance that we get the observations we did when $H_0$ is true**

**The smaller it is, the stronger the evidence against $H_0$**

**If $p < \alpha$, then we reject $H_0$**

Once the choice of test is correct, we can use software to calculate this for us

A beverage factory fills bottles with an advertised volume of 500 milliliters. If the machines underfill the bottles, customers get cheated and regulatory fines follow. If they overfill, the company loses money.

    Null Hypothesis (H0): The mean volume of liquid in the bottles is exactly 500 mL (μ=500).
    Alternative Hypothesis (HA): The mean volume of liquid in the bottles is not 500 mL (μ≠500).

The Test: A two-tailed, one-sample z-test checks a random batch of 100 bottles off the assembly line.

Experiment Outcome: The sample mean is 501.2 mL, with a known machine standard deviation of 4 mL.

```python
from scipy.stats import ttest_ind

import math
from scipy.stats import norm

# Parameters
mu = 500          # Population mean (hypothesized)
sigma = 4        # Population standard deviation (known)
x_bar = 501.2       # Sample mean
n = 100           # Sample size
alpha = 0.05     # Significance level

# Calculate Z-statistic
z_stat = (x_bar - mu) / (sigma / math.sqrt(n))

# Calculate p-value (two-tailed test example)
# For two-tailed: p = 2 * (1 - CDF(|z|))
p_value = 2 * (1 - norm.cdf(abs(z_stat)))

# Calculate critical value
# z_critical = norm.ppf(1 - alpha/2)

print(f"P-Value: {p_value}")

if p_value < alpha:
  print("p is less than alpha, reject the null hypothesis")
else:
  print("The evidence against null hypothesis is not strong enough, do not reject the null hypothesis")
```

- Real life processes always introduce some noise, so the relationship is nondeterministic
- So how do we find out if there is a relation between the variables and the nature of that relation?
- Assume $y = \beta_1 x + \beta_0 + \epsilon$, $\epsilon$ is the noise / random error



## Case study : Identifying the most important factors in car mileage

- First look at a scatter plot of the dependant and independent variables to see if there are any general trends.
- Here we look at the plot of different continuous features vs miles per gallon

```python
!pip install ucimlrepo

from ucimlrepo import fetch_ucirepo
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as plt


# fetch dataset
auto_mpg = fetch_ucirepo(id=9)

# data (as pandas dataframes)
X = auto_mpg.data.features
y = auto_mpg.data.targets

# metadata
print(auto_mpg.metadata)

# variable information
print(auto_mpg.variables)
```

### Examine the dataset

```python
print(X)

print(y)
```

### Missing values in the dataset

- For analysis we need to fill in missing values with a 'sensible' alternative -a constant value, mean, median .etc

- If too many values are missing in a feature, we may need to reconsider it

```python
X['horsepower'] = X['horsepower'].fillna(X['horsepower'].mean())
```

### Exploratory plots

```python
plt.scatter(X["displacement"], y['mpg'])  # displacement, horsepower, weight, acceleration
plt.show()
```

### We attempt to fit a straight line to the data which gives the best fit

- Minimize the total spread of y values from the straight line
- Look at the squared y distances from the line and add them to get Sum of Squared Errors

$SSE = \sum_{i=1}^n (y_i - (\beta_0 + \beta_1 x_i))^2$

The regression line is the one with lowest SSE.

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Fit a linear regression model to the feature and target
reg = LinearRegression().fit(X["displacement"].to_numpy().reshape(-1,1), y['mpg'].to_numpy())

# Find the slope and intercept of the regression line
beta1 = reg.coef_
print("slope", beta1)

beta0 = reg.intercept_
print("intercept", beta0)

# scatter plot of original data
plt.scatter(X["displacement"].to_numpy(), y['mpg'].to_numpy(), label="Original data")
# plot the regression line
X_range = np.arange(50,500)
y_pred = beta0 + (beta1*X_range)
plt.plot(X_range, y_pred, color='red', label="Regression line")
plt.legend()
plt.show()
```

## Squared Correlation

**Proportion of variation in y explained by the regression line**

Variation in $y$

$SS_{yy} = \sum_{i=1}^n (y_i - \bar{y})^2$

$r^2 = 1 - \frac{SSE}{SS_{yy}}$

$r^2 = 1$ : perfect fit

```python
r2 = reg.score(X["displacement"].to_numpy().reshape(-1,1), y['mpg'].to_numpy())
print("squared correlation r2", r2)
```

### Correlation coefficient

**Square root of $r^2$ with the sign of $\beta_0$**

```python
import seaborn as sns

# Combine features and target together to find the correlation of each feature with the target miles per gallon
combined_df = pd.concat([X, y], axis=1)

# 1. Calculate Correlation Matrix of all the features
# Select numeric columns only to ensure accurate calculation
numeric_df = combined_df.select_dtypes(include=['number'])
corr_matrix = numeric_df.corr()

# 2. Create Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix,
    annot=True,          # Display correlation values in cells
    fmt=".2f",           # Format annotations to 2 decimal places
    cmap="coolwarm",     # Use a diverging colormap (blue for negative, red for positive)
    vmin=-1,             # Set min value for color scale
    vmax=1,              # Set max value for color scale
    square=True,         # Make cells square
    linewidths=0.5       # Add lines between cells
)

plt.title("Correlation Heatmap")
plt.show()
```

## Multiple linear regression

- Let us try to use all the continuous variables (most correlated with target) in our data to predict the target

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Fit a linear regression model to the feature and target
reg = LinearRegression().fit(X[["displacement", "cylinders", "horsepower", "weight"]].to_numpy(), y['mpg'].to_numpy())

# Find the slope and intercept of the regression line
beta = reg.coef_
print("slope", beta)

beta0 = reg.intercept_
print("intercept", beta0)
```

```python
# # Look at R2 also for goodness of fit

r2 = reg.score(X[["displacement", "cylinders", "horsepower", "weight"]].to_numpy(), y['mpg'].to_numpy())
print("coefficient of determination r2", r2)
```

```python
y_pred = reg.predict(X[["displacement", "cylinders", "horsepower", "weight"]].to_numpy())

## If prediction is correct original and target values should be equal
X_range = np.arange(10,50)
plt.plot(X_range, X_range, color='red', label="Best result")

# # scatter plot of original target vs the predicted value
plt.scatter( y['mpg'].to_numpy(), y_pred)
plt.xlabel("Predicted value")
plt.ylabel("True value")
plt.legend()
plt.show()
```

## Diagnosing models

- Correlation only tells us that 2 variables are related, it does not say if the relationship is linear

- One simple method : plot residuals $e_i = y_i - \hat{y}_i$ against $y_i$

- A random scatterplot indicates model assumptions are okay

- Any pattern shows there is a problem with model assumptions

```python
y_pred = reg.predict(X[["displacement", "cylinders", "horsepower", "weight"]].to_numpy())
e = y['mpg'].to_numpy() - y_pred

# # scatter plot of original target vs the predicted value
plt.scatter( y['mpg'].to_numpy(), e)
plt.xlabel("y")
plt.ylabel("Residual")
plt.show()
```

### Summary

1.   Explore data by plotting feature vs target to see if they are related

2.   We can see the importance of a feature in predicting the target by looking at the regression weights for that feature

3.   Correlation coefficient tells us if some variables are related to each other

4.   Instead of a single feature, we can use multiple features to predict the target - multiple linear regression

5.   A simple method to diagnose a model that is not predicting correctly - plot the residuals vs the true value

6.   This can highlight if our assumption of linearity is wrong
